```python
import ollama
from datetime import datetime
import re
```

# Utility Function to Sanitize Filenames

```python
def sanitize_filename(text):
    # Remove non-alphanumeric characters and limit to first three words
    words = re.findall(r'\w+', text)
    return '_'.join(words[:3])
```

# Utility Function to Prompt Ollama

```python
def query_llama_with_personality(personality, prompt, model='llama3:8b'):
    """
    Queries the specified Ollama model with a system prompt (personality) and user prompt.
    Saves the output to a Markdown file in the specified format.
    """
    # Prepare messages for Ollama API
    messages = [
        {"role": "system", "content": personality},
        {"role": "user", "content": prompt}
    ]
    
    # Query Ollama
    response = ollama.chat(
        model=model,
        messages=messages
    )
    output_text = response['message']['content']
    
    # Prepare Markdown content
    md_content = f"# prompt: {prompt}\n# personality: {personality}\n\n{output_text}"
    
    # Generate filename
    date_str = datetime.now().strftime("%Y%m%d_%H%M%S")
    fname = f"{sanitize_filename(prompt)}_{date_str}.md"
    
    # Save output
    with open(fname, "w", encoding="utf-8") as f:
        f.write(md_content)
    
    print(f"Output saved to {fname}")
    return output_text
```
