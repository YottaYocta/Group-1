# Persona Testing

There's a debate in the social sciences over language models. Some say they can simulate people, others say they can't. Let's find out!

In this notebook we ask: can language models simulate how a particular person would read a story? Write a story? Annotate a story? 

First, find a story (or two). Then come up with a set of prompts -- they can be like the prompts we've already explored in class, or something entirely new. Determine how the model reacts to the prompt "by default" before giving it a new personality with the System Prompt. Have fun! Can you get the language model to respect the system prompt? Does the model react like how you'd expect it to?

There's no write-up for this testing. Instead, explore the realm of possibilities to get a sense of what is possible.
